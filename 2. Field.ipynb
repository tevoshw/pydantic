# Field

- title: short display name for the field (docs/schema only)
- description: explains what the field is for (docs/schema only)
- default: value used if the field isn't provided
- alias: name used when creating the object, if different from the attribute name
- max_length / min_length: max/min size for str, list, etc
- gt / lt : greater/less than
- pattern: regex the string must match
- decimal_places: max decimal digits allowed
- max_digits: max total digits allowed
- frozen: field can't be changed after object is created
- exclude: field is hidden from model_dump()/model_dump_json(), but still accessible via `.`

> `decimal_places` and `max_digits` only work with `Decimal`, not `float`

> `title` and `description` only appear in the JSON Schema (`.model_json_schema()`), not in `print()` or `model_dump()`

```python
from decimal import Decimal
from pydantic import BaseModel, Field

class Restaurant(BaseModel):
    userID: str = Field(
        title = 'ID',
        alias = 'username',
        default = None,  # If the users don't pass, uses this config as default
        description = 'The name of the User'
    )

    location: str = Field(
        min_length = 10,
        max_length = 30 # Define the limit of caracters to 20 
    )

    money: Decimal = Field(
        decimal_places = 2,
        gt = 0,
        max_digits = 10
    )

    user_password: str = Field(
        exclude = True
    )


restaurant_instance = Restaurant(
    username = 'John Doe',
    location = '123 New York',
    money = 1.99,
    user_password = 'test123'
)

print(restaurant_instance)
```
