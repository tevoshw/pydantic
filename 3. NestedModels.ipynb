````python
from typing import List
from pydantic import BaseModel


class User(BaseModel):
    name: str
    email: str 

class Restaurant(BaseModel):
    name: str
    location: str
    foods: List[str]
    user: User

restaurant_instance = Restaurant(
    name = 'TestsPizzas',
    location = '123 New York',
    foods = ['burger', 'pizza', 'pasta'],
    user = {"name" : 'John Doe', 
             "email" : 'test@gmail.com'} # When use a NESTED MODEL ALWAYS USE A DICT METHOD

)

""" We can use too

    user = {
            'name': 'John Doe',
            'email': 'test@gmail.com'
            }
    and then pass the user

    user = user  
"""

print(restaurant_instance)
````
