#  Computed Field

Create new field with the actual fields


```python

class X(BaseModel):
    first_name: str
    second_name: str

    @computed_field
    @property
    def name(self) -> str:
        return self.first_name + self.second_name

config = X(
    first_name = 'John',
    second_name = 'Doe'
)

config.name # 'JohnDoe

```

````python
from pydantic import BaseModel, computed_field

class Age(BaseModel):
    age1: int
    age2: int

    @computed_field
    @property
    def total_age(self) -> int:
        return self.age1 + self.age2


restaurant_instance = Age(
    age1 = 10,
    age2 = 4
)


print(restaurant_instance)
````
