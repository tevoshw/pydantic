# Field Validator

An pre-processing method in pydantic, that can see only one field per time

```python
from pydantic import BaseModel, field_validator

class Restaurant(BaseModel):
    name: str 

    # Check the 'userID' field 
    @field_validator('name', mode = 'before')
    @classmethod
    def function_verify_name(cls, v):
        if v == 'John Doe':
            return 'Boss'


restaurant_instance = Restaurant(
    name = 'John Doe',
)

print(restaurant_instance)
```

# Model Validator

An pre-processing method in pydantic, that can see more than one field per time

```python
from pydantic import BaseModel, model_validator

class Login(BaseModel):
    name: str 
    user_password: str


    @model_validator(mode='after')
    def admin_login(self):
        if self.name == 'John Doe' and self.user_password == 'Test123':
            self.name = 'Admin'
            return self


restaurant_instance = Login(
    name = 'John Doe',
    user_password = 'Test123'
)

print(restaurant_instance)
```
