```python
from decimal import Decimal
from pydantic.dataclasses import dataclass
from pydantic import Field

@dataclass
class Restaurant:
    userID: str = Field(
        title = 'ID',
        alias = 'username',
        default = None,  # If the users don't pass, uses this config as default
        description = 'The name of the User'
    )

    location: str = Field(
        min_length = 10,
        max_length = 30 # Define the limit of caracters to 20 
    )

    money: Decimal = Field(
        decimal_places = 2,
        gt = 0,
        max_digits = 10
    )

    user_password: str = Field(
        exclude = True
    )


restaurant_instance = Restaurant(
    username = 'John Doe',
    location = '123 New York',
    money = 1.99,
    user_password = 'test123'
)

print(restaurant_instance)
```
