````python
from pydantic import BaseModel

class User(BaseModel):
    x: int
    y: float
````

# BaseModel (Acessing)

```python

config.x
config.y

```

````python
config = User(x = 1, y = 1.2)
print('Test 1')
print(f'Acessing value x: {config.x}')
print(f'Acessing value y: {config.y}')


print(f'=' * 50)
config = User(x = '1', y = '1.3')
print('Test 2')
print(f'Acessing value x: {config.x}')
print(f'Acessing value y: {config.y}')

````
